## Saját megoldások és jegyzetek

```python
import numpy as np
import pandas as pd
```

```python
g = np.random.default_rng()
months = 'Sep Oct Nov Dec Jan Feb Mar Apr May Jun'.split()
s = pd.Series(g.integers(70, 101, 10),
              index=months)
```

```python
print(f'Éves átlag: {s.mean()}')
```

```python
print(s[:5].mean())
print(s[5:].mean())
print(s.iloc[:5].mean())
```

```python
print(s.idxmax())
print(s[s.idxmax()])
```

```python
print(s.round(-1))
```

```python
print(s.sort_values(ascending=False)[:5])
```
